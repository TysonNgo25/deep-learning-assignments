# Kiểm tra có dùng được GPU không

Keras gốc:
```python
from tensorflow.python.client import device_lib
print(device_lib.list_local_devices())
```
TensorFlow tự đặt phép tính lên GPU nếu có. PyTorch thì phải chọn thiết bị và chuyển mô hình, dữ liệu sang bằng `.to(device)`; các notebook sau lấy `device` từ `get_device()` trong `common.py`.

````python
print("CUDA dùng được:", torch.cuda.is_available())
print("Số GPU         :", torch.cuda.device_count())
for i in range(torch.cuda.device_count()):
    p = torch.cuda.get_device_properties(i)
    print(f"  GPU {i}: {p.name}, {p.total_memory / 1024**3:.1f} GB, compute capability {p.major}.{p.minor}")

device = get_device()
````

````python
# Thử một phép nhân ma trận trên thiết bị đã chọn
a = torch.randn(1000, 1000, device=device)
b = a @ a.T
print("Phép tính chạy trên:", b.device, "| shape", tuple(b.shape))
````

## Kết luận

- Có GPU NVIDIA (Colab/Kaggle) thì `device` là `cuda` và dòng trên in tên GPU, giống kết quả gốc "One GPU has been detected!".
- Máy cá nhân dùng GPU AMD nên PyTorch bản CPU báo `cpu`. Các notebook nhẹ (MLP) chạy trên CPU, notebook CNN/LSTM/SegNet chạy trên GPU Colab/Kaggle.
